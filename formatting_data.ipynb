# Code to reformat filter files from nm and % to A and fraction

```python
import pandas as pd
import numpy as np
import glob
```

```python
filter_files = glob.glob('./Inputs/Filters/GeneralFilters/reformat/RC*.txt')
for file in filter_files:
    data = np.loadtxt(file)
    wavelength = data[:,0]
    transmission = data[:,1]
    #convert from nm to A
    #new_wl = wavelength * 10
    #convert from percent to decimal
    new_trans = transmission * 0.01
    #new_data = np.column_stack([new_wl, new_trans])
    new_data = np.column_stack([wavelength, new_trans])
    np.savetxt('./Inputs/Filters/GeneralFilters/reformat/'+file.split('/')[-1], new_data, delimiter=' ')
```

# ROS2 formatting

```python
ros2 = np.genfromtxt('./Inputs/Filters/GeneralFilters/reformat/ROS2.csv', delimiter=',')
wavelength = ros2[:,0]
g = ros2[:,6]
r = ros2[:,7]
i = ros2[:,8]
z = ros2[:,9]

new_wave = wavelength * 10
g_new = g * 0.01
r_new = r * 0.01
i_new = i * 0.01
z_new = z * 0.01

np.savetxt('./Inputs/Filters/GeneralFilters/ROS2_g.dat', np.column_stack([new_wave, g_new]), delimiter=' ')
np.savetxt('./Inputs/Filters/GeneralFilters/ROS2_r.dat', np.column_stack([new_wave, r_new]), delimiter=' ')
np.savetxt('./Inputs/Filters/GeneralFilters/ROS2_i.dat', np.column_stack([new_wave, i_new]), delimiter=' ')
np.savetxt('./Inputs/Filters/GeneralFilters/ROS2_z.dat', np.column_stack([new_wave, z_new]), delimiter=' ')
```

# Extinction correction example

```python
import extinction

#example load in a filter, replace with actual filter for paper/telescope/filter combo:
data = np.loadtxt('./Inputs/Filters/GeneralFilters/DECam_g.dat')
wavelengths = data[:, 0]
transmissions = data[:, 1]

# Calculate the transmission-weighted average
effective_wavelength = np.array([np.sum(wavelengths * transmissions) / np.sum(transmissions)])

# E(B-V) from paper
ebv = 0.105
# Rv = 3.1 for S&F 2011 paper
r_v = 3.1
# or, if they give you an Av value just use that here instead of calculating
a_v = r_v * ebv

# using fitzpatrick 99 law because that is what S&F uses
a_lambda = extinction.fitzpatrick99(effective_wavelength, a_v)

# making a fake test array of magnitudes (replace by reading in actual magnitudes for the paper/telescope/filter combo):
fake_mags_decam_g = np.array([17.45, 17.68, 17.90, 18.34, 19.59])

dust_corr_removed_mags = fake_mags_decam_g - a_lambda[0]
dust_corr_removed_mags
```
